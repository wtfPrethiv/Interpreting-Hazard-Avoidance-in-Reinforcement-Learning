# Behavioural pilots
Single-seed development results do not establish a training-condition effect.

```python
from pathlib import Path
import json
root = Path.cwd() if Path('pyproject.toml').exists() else Path.cwd().parent
paths = sorted((root/'results/manifests').glob('pilot_suite_*.json'))
if not paths:
    raise RuntimeError('Run scripts/run_pilots.py first')
suite = json.loads(paths[-1].read_text())
for run in suite['runs']:
    print(run['run_id'], run['training_steps'], run['readiness'])
```
